```python
from pathlib import Path

IMAGE_DIR = Path(r"C:\Users\SHIVAM\Desktop\New_Projects\Image To Recipie Generation\Image-to-Recipe-Generation\Food Images")

if not IMAGE_DIR.exists():
    raise FileNotFoundError(
        f"Image folder not found: {IMAGE_DIR}"
    )

print("Image folder:", IMAGE_DIR)
```

```python
from collections import Counter

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".webp", ".bmp"
}

image_paths = [
    p for p in IMAGE_DIR.rglob("*")
    if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
]

print("Total image files:", len(image_paths))

print("\nFile extension counts:")
print(Counter(p.suffix.lower() for p in image_paths))

print("\nFirst 10 image paths:")
for path in image_paths[:10]:
    print(path)
```

```python
# Create a normalized image-name lookup.
# Example: dish-name.jpg -> dish-name

image_lookup = {}

for path in image_paths:
    key = path.stem.strip().lower()
    image_lookup.setdefault(key, []).append(path)

# Normalize names in the CSV.
df["image_key"] = (
    df["Image_Name"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
)

# Match each CSV record to a file.
df["image_matches"] = df["image_key"].map(
    lambda key: image_lookup.get(key, [])
)

df["image_found"] = df["image_matches"].map(
    lambda matches: len(matches) > 0
)

df["image_match_count"] = df["image_matches"].map(len)

print("Total recipes:", len(df))
print("Recipes with matching images:", df["image_found"].sum())
print("Recipes without matching images:", (~df["image_found"]).sum())

print(
    "Records with multiple matching files:",
    (df["image_match_count"] > 1).sum()
)

print(
    "Percentage with matching images:",
    round(df["image_found"].mean() * 100, 2), "%"
)
```

```python
# Save a report without copying image data into the CSV.
audit_df = df.drop(columns=["image_matches"])

audit_df.to_csv(
    "../data/dataset_audit.csv",
    index=False
)

# Save the missing image names separately.
missing_images = df.loc[
    ~df["image_found"],
    ["Title", "Image_Name"]
]

missing_images.to_csv(
    "../data/missing_images.csv",
    index=False
)

print("Audit report saved.")
```
